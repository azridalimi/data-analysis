<h1><b>3.2 Using Histogram Plots to Visualize The Distribution of The Field Value for Each Iris Species </b></h1>

```python
# Selecting all the records for Iris-setosa and store it inside setosa DataFrame
setosa=iris_df.loc[iris_df['Species']=='Iris-setosa']

# Selecting all the records for Iris-versicolor and store it inside versi DataFrame
versi=iris_df.loc[iris_df['Species']=='Iris-versicolor']

# Selecting all the records for Iris-Svirginica and store it inside virgi DataFrame
virgi=iris_df.loc[iris_df['Species']=='Iris-virginica']
```

<h2>3.2.2 For each of the flower species DataFrame, we are going to separate the column/field or Pandas Series, so that we can analyse them individually </h2>

```python
# Setosa
setosa_sl=setosa['SepalLength']   # sl = sepal length  sw = sepal width
setosa_sw=setosa['SepalWidth']    # pl = petal length  pw = petal width
setosa_pl=setosa['PetalLength']
setosa_pw=setosa['PetalWidth']

# Versicolor
versi_sl=versi['SepalLength']
versi_sw=versi['SepalWidth']
versi_pl=versi['PetalLength']
versi_pw=versi['PetalWidth']

# Virginica
virgi_sl=virgi['SepalLength']
virgi_sw=virgi['SepalWidth']
virgi_pl=virgi['PetalLength']
virgi_pw=virgi['PetalWidth']
```

<h2>3.2.3 We can plot a histogram to visualise the distribution of petal and sepal length and petal and sepal width for all the flowers </h2>

```python
# 1. Setosa Sepal Length Distribution

from matplotlib import pyplot as plt

plt.hist(setosa_sl,bins=15, color='purple',edgecolor='black')

plt.title('Setosa Sepal Length Value Distribution')
plt.xlabel('Setosa Sepal Length (cm)')
plt.ylabel('frequency')

plt.show()
```

<h3>From the plot we can see that most sepal length for setosa flower lies in the valu between 5.0cm to 5.2cm </h3>

```python
# 2. Setosa Sepal Width  Distribution

from matplotlib import pyplot as plt

plt.hist(setosa_sw,bins=8, color='purple',edgecolor='black')

plt.title('Setosa Sepal Width Value Distribution')
plt.xlabel('Setosa Sepal Width (cm)')
plt.ylabel('frequency')

plt.show()
```

<h3>From the plot we can see that most sepal width for setosa flower lies in the value around 3.0cm to 3.5cm </h3>

```python
# 2. Setosa Sepal Width  Distribution

from matplotlib import pyplot as plt

plt.hist(setosa_pl,bins=9, color='purple',edgecolor='black')

plt.title('Setosa Petal Length Value Distribution')
plt.xlabel('Setosa Petal Length (cm)')
plt.ylabel('frequency')

plt.show()
```

```python
# 2. Setosa Sepal Width  Distribution

from matplotlib import pyplot as plt

plt.hist(setosa_pw,bins=7, color='purple',edgecolor='black')

plt.title('Setosa petal Width Value Distribution')
plt.xlabel('Setosa petal Width (cm)')
plt.ylabel('frequency')

plt.show()
```

<h1><b>4. Using Line Plots to Observe Trends in Average Measurements</b></h1>

```python
features_column=['PetalLengthCm','PetalWidthCm','SepalLengthCm','SepalWidthCm']

# creating a mean_setosa DataFrame by only selecting the features column of the flower
# and then use mean() meathod to calculate the mean value for all feature
mean_setosa=setosa[features_column].mean()
```

```python
# take a peek at the newly created mean_setosa DataFrame
print(mean_setosa)
```

```python
plt.plot(features_column,mean_setosa,marker='o',color='blue',label='setosa')

plt.title('Average Measuremnet for Setosa Species')
plt.xlabel('Feature')
plt.ylabel('Average Measurement')

plt.legend()
plt.show()
```

<h1><b>6.0 Multiple Plot, Visualising All Flower Species</b></h1>

<h2><b>6.1 Using Histogram Plots to Visualize the Distribution Sepal and Width and Sepal and Petal Length</b></h2>

<h3><b>6.1.1 Petal Length Distribution</b></h3>

```python
from matplotlib import pyplot as plt

plt.hist(setosa_pl,bins=10,label='setosa',alpha=0.5)
plt.hist(virgi_pl,bins=10,label='virginica',alpha=0.5)
plt.hist(versi_pl,bins=10,label='versicolor',alpha=0.5)

plt.title('Distribution of Petal Length ')
plt.xlabel('Petal Length')
plt.ylabel('Frequency')

plt.legend()
plt.show()
```

<h2>From the plot, Setosa species can be distinguished from virginica and versicolor species base on its petal length.</h2>

<h3><b>6.1.2 Petal Width Distribution</b></h3>

```python
from matplotlib import pyplot as plt

plt.hist(setosa_pw,bins=6,label='setosa',alpha=0.5)
plt.hist(virgi_pw,bins=6,label='virginica',alpha=0.5)
plt.hist(versi_pw,bins=6,label='versicolor',alpha=0.5)

plt.title('Distribution of Petal Width ')
plt.xlabel('Petal Width')
plt.ylabel('Frequency')

plt.legend()
plt.show()
```

<h2>Petal Length of setosa is also a distinguishable features of Setosa species</h2>

<h3>Plotting multiple Iris flowers on a scatter plot can reveal hidden patterns within the dataset.</h3>

```python
from matplotlib import pyplot as pyplot

plt.scatter(setosa_sl,setosa_pl,color='r',marker='o',label='setosa')
plt.scatter(virgi_sl,virgi_pl,color='g',marker='o',label='virginica')
plt.scatter(versi_sl,versi_pl,color='b',marker='o',label='versicolor')

plt.title('Sepal vs Petal Legth')
plt.xlabel('Sepal Length')
plt.ylabel('Petal Length')

plt.legend()
plt.show()
```

<h2><b>6.3 Multiple Line Plot, Combining the Trend Average Features Measurement For All Iris Species </b></h2>

```python
features_column=['PetalLengthCm','PetalWidthCm','SepalLengthCm','SepalWidthCm']
mean_setosa=setosa[features_column].mean()
mean_virgi=virgi[features_column].mean()
mean_versi=versi[features_column].mean()

plt.plot(features_column,mean_virgi,marker='o',color='blue',label='virginica')
plt.plot(features_column,mean_versi,marker='o',color='green',label='versicolor')
plt.plot(features_column,mean_setosa,marker='o',color='red',label='setosa')

plt.legend()
plt.show()
```
